## 1. O que é um vetor?

Um **vetor** é uma lista ordenada de números. Em Ciência de Dados, usamos vetores o tempo todo: por exemplo, para representar as **notas de um estudante fictício** em quatro avaliações, ou as **medidas de um objeto** (largura, altura, profundidade).

Em Python, representamos um vetor usando um **array do NumPy**.

### Exemplo contextualizado (dados fictícios)

Imagine que "Estudante Fictício A" tirou as seguintes notas em 4 atividades de MECD: 7.0, 8.5, 6.0 e 9.0.


````python
# Exemplo: vetor de notas fictícias de um estudante
notas_estudante_A = np.array([7.0, 8.5, 6.0, 9.0])

print("Vetor de notas:", notas_estudante_A)
print("Dimensão (shape):", notas_estudante_A.shape)
print("Quantidade de elementos:", notas_estudante_A.size)
````

### 📍 Indexação de vetores

Cada posição de um vetor tem um **índice**, começando em **0** (não em 1!). Isso é padrão em Python.

- `notas_estudante_A[0]` → primeira nota
- `notas_estudante_A[1]` → segunda nota
- `notas_estudante_A[-1]` → última nota (índice negativo conta "de trás para frente")


````python
# Indexação: acessando elementos específicos do vetor
primeira_nota = notas_estudante_A[0]
segunda_nota = notas_estudante_A[1]
ultima_nota = notas_estudante_A[-1]

print("Primeira nota:", primeira_nota)
print("Segunda nota:", segunda_nota)
print("Última nota:", ultima_nota)
````

## 2. O que é uma matriz?

Uma **matriz** é uma tabela de números organizada em **linhas** e **colunas**. É exatamente como uma planilha eletrônica ou uma tabela de notas de vários estudantes em várias provas.

A **dimensão** de uma matriz é escrita como `(linhas, colunas)`.

### Exemplo contextualizado (dados fictícios)

Vamos representar as notas fictícias de **3 estudantes** em **4 atividades** de MECD como uma matriz 3×4 (3 linhas = estudantes, 4 colunas = atividades).


````python
# Matriz fictícia de notas: linhas = estudantes, colunas = atividades
notas_turma = np.array([
    [7.0, 8.5, 6.0, 9.0],   # Estudante Fictício A
    [5.5, 6.0, 7.5, 8.0],   # Estudante Fictício B
    [9.0, 9.5, 8.5, 10.0],  # Estudante Fictício C
])

print("Matriz de notas:")
print(notas_turma)
print("\nDimensão (shape):", notas_turma.shape, "-> (linhas=estudantes, colunas=atividades)")
````

### 📍 Indexação de matrizes

Para acessar um elemento de uma matriz, usamos `matriz[linha, coluna]` (lembrando que a contagem começa em 0).

Também podemos acessar uma **linha inteira** ou uma **coluna inteira**:
- `matriz[0, :]` → toda a linha 0 (primeiro estudante)
- `matriz[:, 1]` → toda a coluna 1 (segunda atividade, para todos os estudantes)


````python
# Indexação de matrizes
nota_estudanteB_atividade3 = notas_turma[1, 2]   # linha 1 (Estudante B), coluna 2 (Atividade 3)
todas_notas_estudanteA = notas_turma[0, :]       # linha inteira do Estudante A
todas_notas_atividade1 = notas_turma[:, 0]       # coluna inteira da Atividade 1

print("Nota do Estudante B na Atividade 3:", nota_estudanteB_atividade3)
print("Todas as notas do Estudante A:", todas_notas_estudanteA)
print("Notas de todos na Atividade 1:", todas_notas_atividade1)
````

## 3. Operações simples com vetores e matrizes

### 3.1 Soma e subtração

Somamos ou subtraímos vetores/matrizes **posição por posição** (elemento a elemento). Para isso, as duas estruturas precisam ter a **mesma dimensão**.

**Exemplo contextualizado:** duas provas fictícias de recuperação. Vamos calcular quanto cada estudante fictício **melhorou** entre a primeira tentativa e a segunda tentativa.


````python
# Notas fictícias em duas tentativas de uma prova de recuperação
tentativa_1 = np.array([5.0, 6.5, 4.0])   # Estudantes Fictícios D, E, F
tentativa_2 = np.array([7.0, 6.0, 6.5])

melhora = tentativa_2 - tentativa_1
soma_tentativas = tentativa_1 + tentativa_2

print("Melhora (tentativa 2 - tentativa 1):", melhora)
print("Soma das duas tentativas:", soma_tentativas)
````

### 3.2 Multiplicação por escalar

Multiplicar um vetor ou matriz por um **número (escalar)** significa multiplicar **cada elemento** por esse número.

**Exemplo contextualizado:** o professor decidiu dar um **bônus de 10%** em uma atividade fictícia. Isso é o mesmo que multiplicar o vetor de notas por `1.10`.


````python
# Multiplicação por escalar: bônus de 10% nas notas fictícias
notas_atividade = np.array([6.0, 7.5, 5.0, 8.0])
notas_com_bonus = notas_atividade * 1.10

print("Notas originais:", notas_atividade)
print("Notas com bônus de 10%:", notas_com_bonus)
````

### 3.3 Produto matricial

O **produto matricial** é diferente de multiplicar elemento a elemento! Para multiplicar duas matrizes \(A\) (dimensão \(m \times n\)) e \(B\) (dimensão \(n \times p\)), o **número de colunas de A** precisa ser igual ao **número de linhas de B**. O resultado terá dimensão \(m \times p\).

Cada elemento do resultado é calculado multiplicando **linha por coluna** e somando os produtos.

**Exemplo contextualizado:** uma cantina fictícia da escola vende 3 produtos (suco, sanduíche, fruta). A matriz `quantidade` mostra quantos itens cada estudante fictício comprou, e o vetor `preco` mostra o preço de cada produto. O produto matricial calcula o **total gasto por cada estudante**.


````python
# Produto matricial: quantidade de itens comprados x preço unitário
# Linhas = estudantes fictícios (G, H); Colunas = produtos (suco, sanduíche, fruta)
quantidade = np.array([
    [2, 1, 3],   # Estudante Fictício G
    [1, 2, 0],   # Estudante Fictício H
])

preco = np.array([3.5, 6.0, 2.0])  # preço de: suco, sanduíche, fruta

# Produto matricial (matriz 2x3) x (vetor 3,) = vetor (2,)
total_gasto = quantidade @ preco   # o operador @ representa o produto matricial em NumPy
# Equivalente: np.matmul(quantidade, preco) ou np.dot(quantidade, preco)

print("Quantidade comprada por estudante:\n", quantidade)
print("Preços dos produtos:", preco)
print("Total gasto por estudante:", total_gasto)
````

> ⚠️ **Atenção:** `quantidade * preco` (com asterisco) faria multiplicação elemento a elemento, o que é **diferente** do produto matricial `quantidade @ preco`. Em Ciência de Dados, o produto matricial é essencial — é a base de modelos estatísticos e de aprendizado de máquina.


````python
# Comparando multiplicação elemento a elemento (broadcast) x produto matricial
# Isso é apenas para fins didáticos - observe a diferença de resultado e de forma (shape)

elemento_a_elemento = quantidade * preco  # multiplica cada coluna pelo preço correspondente
print("Elemento a elemento (shape", elemento_a_elemento.shape, "):\n", elemento_a_elemento)

produto_matricial = quantidade @ preco
print("\nProduto matricial (shape", produto_matricial.shape, "):", produto_matricial)
````

## Retomada rápida (5 minutos)

Antes de avançar, vamos relembrar o produto matricial com um mini desafio.


````python
# Retomada: mini desafio de produto matricial
A_retomada = np.array([[1, 2], [3, 4]])
v_retomada = np.array([1, 1])

resultado_retomada = A_retomada @ v_retomada
print("A =\n", A_retomada)
print("v =", v_retomada)
print("A @ v =", resultado_retomada)  # deve ser [3, 7]
````

## 4. Transformações lineares em 2D

Uma **transformação linear** é uma função que "movimenta" pontos ou vetores no plano, mas de forma organizada: ela pode **esticar**, **encolher**, **girar** ou **espelhar** vetores — sempre preservando retas e a origem `(0, 0)`.

Em 2D, uma transformação linear é representada por uma **matriz 2×2**. Para aplicar a transformação a um vetor \(v\), fazemos o produto matricial \(A \cdot v\).

### Exemplo contextualizado: ampliando um desenho no projetor da sala

Imagine um ponto fictício em um desenho digital, representado pelo vetor `(2, 1)`. Vamos aplicar uma matriz de **escala** que dobra o tamanho do desenho.


````python
# Transformação linear: matriz de escala (dobra o tamanho)
ponto_original = np.array([2, 1])

matriz_escala = np.array([
    [2, 0],
    [0, 2],
])

ponto_transformado = matriz_escala @ ponto_original
print("Ponto original:", ponto_original)
print("Ponto após a transformação (escala 2x):", ponto_transformado)
````

````python
# Visualizando a transformação com matplotlib
fig, ax = plt.subplots()

# Vetor original (azul) e vetor transformado (vermelho), saindo da origem (0,0)
ax.quiver(0, 0, ponto_original[0], ponto_original[1], angles='xy', scale_units='xy', scale=1,
           color='blue', label='Ponto original (2,1)')
ax.quiver(0, 0, ponto_transformado[0], ponto_transformado[1], angles='xy', scale_units='xy', scale=1,
           color='red', label='Ponto transformado (4,2)')

ax.set_xlim(-1, 5)
ax.set_ylim(-1, 5)
ax.axhline(0, color='gray', linewidth=0.5)
ax.axvline(0, color='gray', linewidth=0.5)
ax.set_title("Transformação linear: escala 2x")
ax.set_xlabel("eixo x")
ax.set_ylabel("eixo y")
ax.legend()
ax.grid(True, linestyle='--', alpha=0.5)
ax.set_aspect('equal')
plt.show()
````

## 5. Combinação linear

Uma **combinação linear** de vetores é o resultado de multiplicar cada vetor por um número (escalar) e depois somar tudo.

Se temos os vetores \(v_1\) e \(v_2\), uma combinação linear é:

\[
c_1 \cdot v_1 + c_2 \cdot v_2
\]

onde \(c_1\) e \(c_2\) são números quaisquer.

### Exemplo contextualizado: misturando ingredientes fictícios de uma receita

Imagine dois vetores fictícios que representam "unidades de sabor" de dois sucos: suco A = `(2, 0)` (muito doce, pouco ácido) e suco B = `(0, 3)` (pouco doce, muito ácido). Uma combinação linear representa **misturar quantidades diferentes** desses sucos.


````python
# Combinação linear de dois vetores fictícios
suco_A = np.array([2, 0])
suco_B = np.array([0, 3])

# Mistura: 2 partes de suco A + 1 parte de suco B
mistura = 2 * suco_A + 1 * suco_B
print("Suco A:", suco_A)
print("Suco B:", suco_B)
print("Mistura (2A + 1B):", mistura)
````

````python
# Visualizando a combinação linear
fig, ax = plt.subplots()

ax.quiver(0, 0, suco_A[0], suco_A[1], angles='xy', scale_units='xy', scale=1, color='orange', label='Suco A')
ax.quiver(0, 0, suco_B[0], suco_B[1], angles='xy', scale_units='xy', scale=1, color='green', label='Suco B')
ax.quiver(0, 0, mistura[0], mistura[1], angles='xy', scale_units='xy', scale=1, color='purple', label='Mistura (2A+1B)')

ax.set_xlim(-1, 6)
ax.set_ylim(-1, 6)
ax.axhline(0, color='gray', linewidth=0.5)
ax.axvline(0, color='gray', linewidth=0.5)
ax.set_title("Combinação linear de vetores")
ax.legend()
ax.grid(True, linestyle='--', alpha=0.5)
ax.set_aspect('equal')
plt.show()
````

## 6. Base e dependência linear

- Um conjunto de vetores forma uma **base** do plano 2D quando conseguimos escrever **qualquer ponto do plano** como combinação linear desses vetores, e nenhum vetor do conjunto é "redundante".
- Dois vetores são **linearmente dependentes** quando um deles é apenas um múltiplo do outro (estão na mesma direção/reta). Nesse caso, **não** formam uma base do plano — eles não conseguem "cobrir" todas as direções.
- Dois vetores são **linearmente independentes** quando não estão na mesma direção. Se forem 2 vetores independentes no plano 2D, eles formam uma base.

### Como verificar rapidamente em 2D?

Para vetores 2D \(v_1 = (a, b)\) e \(v_2 = (c, d)\), calculamos o **determinante**:

\[
\det = a \cdot d - b \cdot c
\]

- Se \(\det \neq 0\): os vetores são **linearmente independentes** (formam uma base do plano).
- Se \(\det = 0\): os vetores são **linearmente dependentes** (estão na mesma reta, não formam uma base).


````python
# Exemplo: verificando dependência/independência linear com determinante

def sao_independentes(v1, v2):
    '''Retorna True se v1 e v2 forem linearmente independentes (formam base do plano 2D).'''
    matriz = np.array([v1, v2])
    det = np.linalg.det(matriz)
    return not np.isclose(det, 0), det

# Caso 1: vetores independentes (direções diferentes)
v1, v2 = np.array([1, 0]), np.array([0, 1])
independentes, det = sao_independentes(v1, v2)
print(f"v1={v1}, v2={v2} -> determinante={det:.2f} -> independentes? {independentes}")

# Caso 2: vetores dependentes (v2 é múltiplo de v1, mesma direção)
v3, v4 = np.array([2, 1]), np.array([4, 2])
dependentes, det2 = sao_independentes(v3, v4)
print(f"v3={v3}, v4={v4} -> determinante={det2:.2f} -> independentes? {dependentes}")
````

````python
# Visualizando os dois casos lado a lado
fig, axs = plt.subplots(1, 2, figsize=(11, 5))

# Caso independente
axs[0].quiver(0, 0, v1[0], v1[1], angles='xy', scale_units='xy', scale=1, color='blue', label='v1')
axs[0].quiver(0, 0, v2[0], v2[1], angles='xy', scale_units='xy', scale=1, color='red', label='v2')
axs[0].set_xlim(-1, 3); axs[0].set_ylim(-1, 3)
axs[0].set_title("Independentes (formam base)")
axs[0].axhline(0, color='gray', linewidth=0.5); axs[0].axvline(0, color='gray', linewidth=0.5)
axs[0].legend(); axs[0].grid(True, linestyle='--', alpha=0.5); axs[0].set_aspect('equal')

# Caso dependente
axs[1].quiver(0, 0, v3[0], v3[1], angles='xy', scale_units='xy', scale=1, color='blue', label='v3')
axs[1].quiver(0, 0, v4[0], v4[1], angles='xy', scale_units='xy', scale=1, color='red', label='v4')
axs[1].set_xlim(-1, 5); axs[1].set_ylim(-1, 3)
axs[1].set_title("Dependentes (mesma direção)")
axs[1].axhline(0, color='gray', linewidth=0.5); axs[1].axvline(0, color='gray', linewidth=0.5)
axs[1].legend(); axs[1].grid(True, linestyle='--', alpha=0.5); axs[1].set_aspect('equal')

plt.tight_layout()
plt.show()
````
